## IRIS FLOWER CLASSIFICATION USING K-NEAREST NEIGHBORS (KNN)

#### Import pandas library

```python
import pandas as pd
```

#### Import the built-in Iris dataset from Scikit-Learn

```python
from sklearn.datasets import load_iris
```

#### Import function to split dataset into training and testing sets

```python
from sklearn.model_selection import train_test_split
```

#### Import KNN classification algorithm

```python
from sklearn.neighbors import KNeighborsClassifier
```

#### Import accuracy calculation function

```python
from sklearn.metrics import accuracy_score
```

### STEP 1: LOAD THE IRIS DATASET

#### Load the Iris dataset

```python
iris = load_iris()
```

### STEP 2: CREATE A DATAFRAME

#### Convert dataset into tabular form

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
```

#### Add target(species) column

```python
df['species'] = iris.target
```

#### Display first 5 rows

```python
print("First 5 Rows of Dataset:")
print(df.head())
```

### STEP 3: SEPARATE FEATURES AND TARGET

#### Remove species column because species is what we want to predict.

```python
X = df.drop('species', axis=1)
```

#### Species column becomes target variable

```python
y = df['species']
```

### STEP 4: SPLIT DATA INTO TRAINING AND TESTING SETS

#### train_test_split divides data into two parts:

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,                      # Input features
    y,                      # Output labels
    test_size=0.2,          # 20% data for testing
    random_state=42         # Fixed random split
)
```

#### Print sizes

```python
print("\nTraining Samples:", len(X_train))
print("Testing Samples:", len(X_test))
```

### STEP 5: CREATE KNN MODEL

#### Create KNN classifier

```python
model = KNeighborsClassifier(n_neighbors=3)
```

### STEP 6: TRAIN THE MODEL

#### fit() teaches the model using training data

```python
model.fit(X_train, y_train)
print("\nModel Training Completed Successfully!")
```

### STEP 7: TEST THE MODEL

#### Predict species for unseen testing data

```python
y_pred = model.predict(X_test)
```

### STEP 8: CALCULATE ACCURACY

```python
accuracy = accuracy_score(y_test, y_pred)
print("\nModel Accuracy:", accuracy)
```

#### Convert to percentage

```python
print("Accuracy Percentage:", accuracy * 100, "%")
```

### STEP 9: PREDICT A NEW FLOWER

```python
new_flower = pd.DataFrame([[5.1, 1.5, 3.4, 0.2]], columns = iris.feature_names)
```

#### Predict species

```python

prediction = model.predict(new_flower)
print("\nPredicted Class Number:", prediction[0])
```

### STEP 10: CONVERT CLASS NUMBER TO SPECIES NAME

```python
species_name = iris.target_names[prediction[0]]
print("Predicted Species:", species_name)
```
